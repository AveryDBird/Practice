```python
from datetime import *
today = date.today()
print(today)
print(today.year)
print(today.month)
print(today.day)
print(today.weekday())
print("the current time is", today.ctime())
```

```python
now = datetime.now()
print(now.strftime("%d/%m/%Y %H:%M:%S"))
print(now.strftime("the year is %Y"))
```

```python
print(timedelta(days=365,hours= 5, minutes= 1))
print("in one year itll be" ,now + timedelta(days = 365))
```

```python
today = date.today()
afd = date(today.year, 4,1)
if afd < today:
  print(f"AFD passed {(today-afd).days} ago")
```
