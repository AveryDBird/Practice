```python
today = date.today()
afd = date(today.year, 4,1)
if afd < today:
  print(f"AFD passed {(today-afd).days} ago")
  afd = afd.replace(year= today.year + 1)
  time_to_afd = afd-today
print(f"only {time_to_afd.days} till april fools day")
```

```python
print("team meetings will be")
for m in range(1,13):
  cal = calendar.monthcalendar(2026,m)
  weekone = cal[0]
  weektwo = cal[1]
  if weekone[calendar.FRIDAY] != 0:
    meeting = weekone[calendar.FRIDAY]
  else:
    meeting = weektwo[calendar.FRIDAY]
  print(f"{calendar.month_name[m]} : {meeting}")
```

```python
import calendar


show_expected_result = False
show_hints = False

def count_days(year, month, whichday):
    result = 0
    _, num_days = calendar.monthrange(year,month)
    for day in range(1, num_days +1):
        if calendar.weekday(year,month,day) == whichday:
            result +=1

    return result
```
