```python
import os
from os import path

show_expected_result = False
show_hints = False

def file_info():
    filesize = 0
    depscontent = os.listdir("deps")
    for i in depscontent:
        if i.endswith(".txt"):
            full_path = os.path.join("deps",i)
            filesize = filesize + os.path.getsize(full_path)
    return filesize
```
